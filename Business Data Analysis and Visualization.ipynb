```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"c:\Users\User\thanush\2. Finance_Analytics_Dataset (1).csv")
```

```python
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Category', order=df['Category'].value_counts().index)
plt.title('Number of Transactions by Category')
plt.xlabel('Category')
plt.ylabel('Transaction Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df.info
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='PaymentMode', order=df['PaymentMode'].value_counts().index)
plt.title('Number of Transactions by Payment Mode')
plt.xlabel('Payment Mode')
plt.ylabel('Transaction Count')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.countplot(data=df, x='TransactionType')
plt.title('Number of Transactions by Type')
plt.xlabel('Transaction Type (Debit/Credit)')
plt.ylabel('Transaction Count')
plt.tight_layout()
plt.show()
```

```python
sns.barplot(data=df, x='TransactionType', y='Amount', estimator=sum, ci=None)
plt.title('Total Amount Comparison: Debit vs Credit')
plt.xlabel('Transaction Type')
plt.ylabel('Total Amount')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='CustomerSegment', y='Amount', estimator=sum, ci=None)
plt.title('Total Spending Pattern by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Spending Amount')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 7))
payment_counts = df['PaymentMode'].value_counts()
plt.pie(payment_counts, labels=payment_counts.index, autopct='%1.1f%%', startangle =90)
plt.title('Payment Mode Usage Distribution')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Category', y='Amount')
plt.title('Distribution of Transaction Amount by Category')
plt.xlabel('Category')
plt.ylabel('Transaction Amount')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
